```python
import pandas as pd

data = {
"product_id": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
"price": [25000, 45000, 15000, 60000, 35000, 20000, 55000, 30000, 40000, 50000],
"quantity": [4, 2, 6, 1, 3, 5, 2, 7, 3, 2],
"discount": [5, 10, 0, 15, 5, 10, 5, 0, 10, 15]
}


sales_df = pd.DataFrame(data)
sales_df
```

```python
#1. New column named total_sales

total_sales = sales_df['price'] * sales_df['quantity']
total_sales

sales_df['total_sales'] = total_sales
sales_df
```

```python
#2. Average total sales across all products.

avg_total_sales = total_sales.mean()
print('Average sales is', avg_total_sales)

avg_total_sales = sales_df['total_sales'].mean()
print('Average sales is', avg_total_sales)
```

```python
#3. maximum and minimum values in the total_sales column.

max_total_sales = total_sales.max()
print('Maximum sales is', max_total_sales)

min_total_sales = total_sales.min()
print('Minimum sales is', min_total_sales)


max_total_sales = sales_df['total_sales'].max()
print('Maximum sales is', max_total_sales)

min_total_sales = sales_df['total_sales'].min()
print('Minimum sales is', min_total_sales)
```

```python
#4.  product_id of the product with the highest total_sales

highest_sales = sales_df['total_sales'].max()
print('Highest sales is', highest_sales)

highest_quantity = sales_df['total_sales'].idxmax()
print('Highest quantity is', highest_quantity)

highest_productid = sales_df.loc[sales_df['total_sales'].idxmax(), 'product_id']
print('Highest product_id is', highest_productid)
```

```python
#5. all rows for products whose total_sales are greater than the average total sales.

product_id = sales_df['total_sales'] > avg_total_sales
print(product_id)

product_id = sales_df[sales_df['total_sales'] > avg_total_sales]
print(product_id)
```

```python
#6. all products where quantity sold is greater than 4.

qty_sold = sales_df['quantity'] > 4
print(qty_sold)

qty_sold = sales_df[sales_df['quantity'] > 4]
print(qty_sold)
```

```python
#7. price after applying the discount percentage and store the result in a new column named discounted_price

discounted_price = sales_df['price'] - (sales_df['price'] * sales_df['discount'] / 100)
discounted_price

sales_df['discounted_price'] = discounted_price
sales_df
```

```python
#13. Calculate the mean, median, variance, and standard deviation of the total_sales column

#mean
mean_tsales = sales_df['total_sales'].mean()
print('Mean total sales is', mean_tsales)

#median
median_tsales = sales_df['total_sales'].median()
print('Median total sales is', median_tsales)

#variance
var_tsales = sales_df['total_sales'].var()
print('Variance total sales is', var_tsales)

#standard deviation
std_tsales = sales_df['total_sales'].std()
print('Standard deviation total sales is', round(std_tsales, 2))
```
